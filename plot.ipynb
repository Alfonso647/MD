### 1.2 Creación de una gráfica

Vamos a crear nuestra primera gráfica empleando las funciones que nos ofrece 
`Matplotlib`. Para crear una gráfica, usamos la función `plot()`. Esta función recibe como parámetro los puntos que se van a representar, en este caso recibe dos argumentos que se refieren a el ejex y ejey. Además de `plot()`, podemos usar varias funciones para personalizar y mostrar nuestra gráfica. Por ejemplo, tenemos:
- plt.figure(): Esta función define el tamaño de nuestra gráfica. Si no usamos esta función, se crea un tamaño de gráfica por defecto
- plt.show(): Esta función es necesaria si queremos mostrar por pantalla nuestro plot 

Vamos a ver algunos ejemplos creando algunas gráficas simples:

```python
eje_x = [1, 2, 3, 4, 5]
eje_y = [3, 3, 3, 5, 7]

plt.figure(figsize=(10, 5)) 
plt.plot(eje_x, eje_y) 
plt.show() 
```

En el ejemplo anterior, hemos creado una función lineal, pero esta librería nos ofrece diferenes formas de mostrar las gráficas. Un ejemplo es usando la función `scatter()`. Esta función produce un gráfico de dispersión y toma los mismos parámetros que `plot()`, que son el ejex y el ejey.

Vamos a crear un gráfico de dispersión a modo de ejemplo:

```python
eje_x = [1, 2, 3, 4, 5]
eje_y = [3, 3, 3, 5, 7]

plt.figure(figsize=(10, 5)) 
plt.scatter(eje_x, eje_y) 
plt.show() 
```

Vamos a seguir explorando las distintas formas de representación de gráficas que nos ofrece la librería. En este caso, vamos a estudiar como graficar en 3D. Para ello necestiamos nuevamente la función `plot()`, pero en lugar de tomar dos arguementos, toma tres. Estos se refierene al ejex, ejey, y ejez para añadir profundiad y crear las tres dimensiones. 

Vamos a crear un ejemplo de un gráfico en 3D:

```python
np.random.seed(42)
fig = plt.figure(figsize=(10, 5)) 
ax = fig.add_subplot(111, projection='3d') 

x = np.random.standard_normal(100) 
y = np.random.standard_normal(100) 
z = np.random.standard_normal(100) 

ax.scatter(x, y, z) 
plt.show() 
```

Otra opción de la que disponemos para mostrar datos es graficando una gráfica de barras. Para ello no usamos plot, si no que usamos la función `bar()`. Esta función nuevamente recibe como argumentos los datos del ejex y ejey. 

Vamos a crear un ejemplo de un gráfico de barras emplenado las herramientas comentadas:

```python
eje_x = [1, 2, 3, 4, 5, 6, 7, 8, 9]
eje_y = [0, 2, 4, 8, 10, 8, 4, 2, 0]

plt.figure(figsize=(10, 5)) 
plt.bar(eje_x, eje_y) 
plt.show() 
```
