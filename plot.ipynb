## 2 Creación de una gráfica

Vamos a crear nuestra primera gráfica empleando las funciones que nos ofrece 
`Matplotlib`. Para crear una gráfica, usamos la función `plot()`. Esta función recibe como parámetro los puntos que se van a representar, en este caso recibe dos argumentos que se refieren a el ejex y ejey. Además de `plot()`, podemos usar varias funciones para personalizar y mostrar nuestra gráfica. Por ejemplo, tenemos:
- plt.figure(): Esta función define el tamaño de nuestra gráfica. Si no usamos esta función, se crea un tamaño de gráfica por defecto
- plt.show(): Esta función es necesaria si queremos mostrar por pantalla nuestro plot 

Vamos a ver algunos ejemplos creando algunas gráficas simples:

````python
eje_x = [1, 2, 3, 4, 5]
eje_y = [3, 3, 3, 5, 7]

plt.figure(figsize=(10, 5)) 
plt.plot(eje_x, eje_y) 
plt.show() 
````

En el ejemplo anterior, hemos creado una función lineal, pero esta librería nos ofrece diferenes formas de mostrar las gráficas. Un ejemplo es usando la función `scatter()`. Esta función produce un gráfico de dispersión y toma los mismos parámetros que `plot()`, que son el ejex y el ejey.

Vamos a crear un gráfico de dispersión a modo de ejemplo:

````python
eje_x = [1, 2, 3, 4, 5]
eje_y = [3, 3, 3, 5, 7]

plt.figure(figsize=(10, 5)) 
plt.scatter(eje_x, eje_y) 
plt.show() 
````

Vamos a seguir explorando las distintas formas de representación de gráficas que nos ofrece la librería. En este caso, vamos a estudiar como graficar en 3D. Para ello necestiamos nuevamente la función `plot()`, pero en lugar de tomar dos arguementos, toma tres. Estos se refierene al ejex, ejey, y ejez para añadir profundiad y crear las tres dimensiones. 

Vamos a crear un ejemplo de un gráfico en 3D:

````python
np.random.seed(42)
fig = plt.figure(figsize=(10, 5)) 
ax = fig.add_subplot(111, projection='3d') 

x = np.random.standard_normal(100) 
y = np.random.standard_normal(100) 
z = np.random.standard_normal(100) 

ax.plot(x, y, z) 
plt.show() 
````

Otra opción de la que disponemos para mostrar datos es graficando una gráfica de barras. Para ello no usamos plot, si no que usamos la función `bar()`. Esta función nuevamente recibe como argumentos los datos del ejex y ejey.

````python
eje_x = [1, 2, 3, 4, 5, 6, 7, 8, 9]
eje_y = [0, 2, 4, 8, 10, 8, 4, 2, 0]

plt.figure(figsize=(10, 5)) 
plt.bar(eje_x, eje_y) 
plt.show() 
````

Finalmente, vamos a estudiar como graficar un histograma. Para ello, `Matplotlib` nos ofrece la función `hist()`, la cual recibe como argumentos los datos. Un histograma es una
representación gráfica de la distribución de un conjunto de datos numéricos y continuos, organizada mediante barras verticales, donde se muestra la frecuencia con la que suceden los valores dentro de intervalos. Luego esta función recibe dos argumentos, que son dichos datos e intervalos.

Vamos a crear un histograma a modo de ejemplo:

````python
np.random.seed(2)
data = np.random.randn(100) # Create random data

plt.hist(data, bins=45) # Create a histogram with 30 bins. Here bins are the number of intervals in the data, and it is a parameter of the histogram
                        # if bins is too small, the histogram will be too smooth, and if bins is too large, the histogram will be too spiky

plt.show() # Show the plot
````

### 3.1 Añadir un título

El recurso más importante es ser capaz de nombrar nuestra gráfica. Es el primer paso a la hora de presentar los plots a los demás para que entiendad que es lo que estamos graficando exactamente. Para ello, usamos la función `title()` que recibe como argumento   el titulo de la gráfica. 

Vamos a poner varios ejemlos en los cuales nombramos a nuestras gráficas:

````python
# Generar 100 puntos entre -10 y 10
x = np.linspace(0, 30, 100) 
y = np.sin(x) 

plt.plot(x, y)
plt.title('Función seno')
plt.show() 
````

### 3.2 Añadir etiquetas

Otro recurso sumamente importante es añadir nombres a los ejes de la figura que estemos mostrando. Es importante mostar que estamos comparndo con que para que las gráficas sean entendibles. Para esto, usamos las siguientes funciones:

- `xlabel()`: Para el ejex, recibe como parámetro la etiqueta del ejex
- `ylabel()`: Para el ejex, recibe como parámetro la etiqueta del ejey

Vamos a crar otro ejemplo de gráfica para mostar el funcionamieto de estas funciones:

````python
# Generar 100 puntos entre -10 y 10
x = np.linspace(0, 30, 100) 
y = np.cos(x) 

plt.plot(x, y)
plt.title('Función coseno')
plt.xlabel('x')
plt.ylabel('cos(x)')
plt.show() 
````

### 3.3 Añadir una leyenda

Es también muy importante definir las leyendas de las gráficas. Sobretodo cuando  en una misma gráfica mostramos vamos a mostrar varias representaciones. La leyenda de un gráfico es un elemento explicativo (suele ser un cuadro pequeño) que identifica qué información representa cada linea. Para implementar las leyendas usamos la función: `legend()`. En el caso de las leyendas, especificamos el nombre de la función que van a representar en el plot de la misma añadiendole un parámetro:

- label="Nombre de la etiqueta"

Volviendo a los ejemplos anteriores, vamos a mostrar la función seno y coseno juntas. Para mostrar plots en la misma gráfica, simplemente usamos la función `plot()` para la función coseno(x) y para seno(x). 

````python
x = np.linspace(0, 30, 100) 
y_sin = np.sin(x)
y_cos = np.cos(x) 

plt.figure(figsize=(10, 5)) 

# función seno
plt.plot(x, y_sin, label='seno')
plt.xlabel('x')
plt.ylabel('sin(x)')

# Función coseno
plt.plot(x,y_cos, label='coseno')
plt.xlabel('x')
plt.ylabel('cos(x)')

# Representamos la gráfica
plt.title('Representación seno vs coseno')
plt.legend()
plt.show() 
````

### 3.4 Personzalizar el color, estilo de línea y estilo de marcador

Notamos como en el ejemplo anterior, al mostar dos gráficas y no le especificamos nada más `Matplotlib` cambia los colores del segundo plot con respecto al primero. Esto es para que se distingan y por defecto le da al primer plot el color azul y al segundo el color naranja. En este apartado vamos a estudiar como modificar el estilo de linea, color... Para ello, especificamos estas personalizaciones dentro de la propia función `plot()`:

```python
plt.plot(x, y, color = '', linestyle= '', marker= '')
```
Vemos como tenemos que darle valor a los siguientes argumentos:
- color -> Color de línea
- linestyle -> Estilo de línea
- marker -> Estilo de marcador

Vamos a poner varios ejemplos de gráficas con nuestros propios estilos. Para mostrar curvas más visuales vamos a emplear gráficas polares:
-> En lugar de usar coordenadas X,Y; usamos un ángulo ``θ`` y una distancia ``r`` desde el centro. Vamos a seguir los siguientes pasos:

- Para calcular θ:
-> Calculamos puntos desde 0 a 2π puesto que queremos cubrir un giro completo

- Para calcular r
-> Depende de que curva queremos representar

- Convertimos a coordenadas cartesianas
-> Una vez que tenemos las distancias r calculadas, vamos a pasarlas a coordenadas cartesianas con el objetivo de una mejor representación. En este caso : \
1. x = r * cos(θ)    # Proyección horizontal
2. y = r * sin(θ)    # Proyección vertical

Una vez tenemos x e y, graficamos como lo hemos estado haciendo hasta ahora

Curvas de referencia: \
https://matematicasiesoja.wordpress.com/wp-content/uploads/2020/04/curvasmaravillosas.pdf

````python
import numpy as np
import matplotlib.pyplot as plt

angulo = np.linspace(0, 2*np.pi, 1000)
r = np.sin(6 * angulo)

# Pasamos a coordenadas cartesianas
x = r * np.cos(angulo)
y = r * np.sin(angulo)

plt.plot(x, y, color='pink', linestyle='-.', marker='.')
plt.title('Folio')
plt.show()
````

Vemos como hemos usado :
- color = 'pink' (le da el color rosa)
- linestyle = '-.' (los puntos se unen con guión y punto)
- marker = '.' (cada punto se representa con un punto)

Vamos a poner otro ejemplo quitando el marcador para que se aprecie más el parámetro de `linestyle`:

````python
angulo = np.linspace(0, 2*np.pi, 1000)
r = np.sin(4 * angulo)

# Pasamos a coordenadas cartesianas
x = r * np.cos(angulo)
y = r * np.sin(angulo)

plt.plot(x, y, color='blue', linestyle='-.')
plt.title('Folio')
plt.show()
````

## 4. Múltiples gráficas

### 4.1 Versión en bruto

`Matplotlib` nos permite crear varias gráficas dentro de una misma figura. Esto lo conseguimos con la función `subplot()`, la cual recibe los siguentes parámetros:
- Número de filas
- Número de columnas
- Índice del grafo

Vamos a ver un ejemplo:

````python
# Primer plot
angulo = np.linspace(0, 5*np.pi, 100)
r = 2 * np.exp(0.3*angulo)

# Pasamos a coordenadas cartesianas
x = r * np.cos(angulo)
y = r * np.sin(angulo)


plt.subplot(1,2,1)
plt.plot(x, y, color='red', linestyle=':')
plt.title('Espiral logarítmica')


# Segundo plot
r = 0.8 * angulo
# Pasamos a coordenadas cartesianas
x = r * np.cos(angulo)
y = r * np.sin(angulo)

plt.subplot(1,2,2)
plt.plot(x, y, color='green', linestyle='--')
plt.title('Espiral de Arquímedes')
plt.show()
````

Vemos como hemos crado dos espirales. Para ello hemos tenido que aumentar el valor del ángulo con el objetivo de que curbiera mas de una sola vuelta completa. En este caso, hemos usado subplots de la siguiente forma:

- subplot(1,2,1)
- subplot(1,2,2)

Esto nos crea una figura de 1 fila y 2 columnas como podemos observar en el resultado. En el siguiente apartado vamos a esutudiar com ocrear múltiples gráficos usando objetos

### 4.2 Versión formateada

Usar objetos para crear diversos gráficos es una de las formas mas óptimas de hacerlo. Vamos a explicar el proceso de forma detallada:

El primer paso es obtener los objetos: `fig` y `ax` de la siguiente forma:
```python
fig, ax = plt.subplots(1, 2, figsize=(10, 5)) # Create a figure and 2 axes. 
```
En este caso, tenemos una sola figura y dos ejes (lugares individuales donde se dibujan cada gráfico) es decir que vamos a dibujar dos gráficas dentro de una figura.

Para personalizar cada gráfico, usamos el objeto ax y su índice, es decir, si queremos acceder al primer "mini"-gráfico usamos: ax[0]. Luego, `Matplotlib` nos ofrece una serie de funciones para modificar cada gráfica:
- .plot -> Crear el gráfico (usando plot)
- .set_title -> Añadir un título
- .set_xlabel & .set_ylabel -> Añadir título a los ejes
- .legend -> Mostrar las leyendas

Vamos a mostrar un ejemplo de como usar objetos para crear varias gráficas:

````python
angulo = np.linspace(1, 15*np.pi, 100)

# Creamos los objetos
fig, ax = plt.subplots(1, 2, figsize=(10, 5))


# PRIMER PLOT
r = 1 * (np.sqrt(angulo))
# Pasamos a coordenadas cartesianas
x = r * np.cos(angulo)
y = r * np.sin(angulo)

ax[0].plot(x, y, label='Femat', color='Orange', linestyle='-.') 
ax[0].set_title('Espiral de Fermat') 
ax[0].set_xlabel('eje x')
ax[0].set_ylabel('eje y')
ax[0].legend()

# SEGUNDO PLOT
r = 1 / (np.sqrt(angulo))
# Pasamos a coordenadas cartesianas
x = r * np.cos(angulo)
y = r * np.sin(angulo)

ax[1].plot(x, y, label='Lituus', color = 'Black', linestyle='--') 
ax[1].set_title('Espiral de Lituus')
ax[1].set_xlabel('eje x')
ax[1].set_ylabel('eje y')
ax[1].plot(y-1, x+1, label='Lituus', color = 'red', linestyle=':') 
ax[1].legend()
plt.show()
````

Otra función que nos permite esta librería es la de guardar nuestras gráficas en distintos formatos de archivos. Para ello, usamos la función `savefig()`. Esta función recibe como argumento el nombre que va a tener el archivo guardado y en qur formato (jpg, png...) Vamos a guardar una gráficaen un archivo llamado `plot.png`:

````python
angulo = np.linspace(1, 5*np.pi, 100)

r = 4 * (1 - np.cos(angulo))
# Pasamos a coordenadas cartesianas
x = r * np.cos(angulo)
y = r * np.sin(angulo)

plt.figure(figsize=(10, 5)) 
plt.plot(x, y, label='Cardioide1', color='red')
plt.plot(x-11, y, label='Cardioide2', color='yellow')
plt.title('Cardiodide')
plt.xlabel('ejex')
plt.ylabel('ejey')
plt.axis('equal')  # AJustamos dimensiones de los ejes
plt.legend()
# Guardamos la figura
plt.savefig('plot.png')
````

### 5.2 Cración de una gráfica

Vamos a esutidar que opciones nos permite `Seaborn` para crear diferentes gráficas. Para crear una gráfica simple, tenemos la función `lineplot()`. Esta función sería el equivalente a `plot()` de `Matplotlib`. La función tiene dos argumentos:
- Eje x
- Eje y

`Seaborn` representa el eje y frente al eje x

Vamos a crear nuestra primera gráfica con esta biblioteca:

````python
x = np.linspace(-10, 10, 20)
plt.figure(figsize=(10,5))
y = x**2

sns.lineplot(x=x, y=y, color='Green', linestyle='-.', marker='d', label='Función')
plt.title('Función cuadrática')
plt.legend()
plt.show()
````

Además, `Seaborn` nos permite crar gráficas de barras, diagramas de dispersión, histogramas... Vamos a listar algunos funciones ofrecidas por esta biblioteca y luego vamos a mostrar varios ejemplos:

- barplot
- scatterplot
- histplot

Todas estas funciones representan el eje_y frente al eje_x. Vamos a poner varios ejemplos:

##### BARPLOT

````python
categorias = ['A', 'B', 'C', 'D']
valores = [10, 23, 17, 9]

# Graficar
sns.barplot(x=categorias, y=valores, linewidth=3, edgecolor='black', capsize=0.2)
plt.title('Barplot')
plt.show()
````

##### SCATTERPLOT

````python
x = np.random.rand(50) 
y = np.random.rand(50) 
sns.scatterplot(x=x, y=y, color='black', s=80, edgecolor = 'red')
plt.title('Scatterplot')
plt.show
````

##### HISTPLOT

````python
datos = np.random.normal(0, 1, 1000)
sns.histplot(datos, bins=20, kde=True, color='purple')
plt.title('Histplot')
plt.show()
````
