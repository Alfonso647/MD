## 1. Creación de matrices

En esta sección vamos a definir como se crean los arrays usando la librería Numpy. Los arrays son estructuras de datos que almacenan una pila de elementos del mismo tipo. Para obtener estos arrays, usamos la función que nos proporciona numpy: array. Esta función recibe una lista y devuelve su array.

```python
array = np.array([0,5,10,20,25,30])    # Creamos el array
print(array)    # Mostramos el array por pantalla 
```

Para formar una matriz, solo tenemos que crear el array (con el mismo método explicado arriba) usando una lista que contiene listas. Vamos a formular un ejemplo para ver su funcionamiento

```python
array = np.array([[9,5,2],[1,1,1],[3,2,4]])
print(array)
```

Otra función que nos permite esta librería es la de crear arrays con números aleatorios. Para este función usamos random.rand(). Esta función recibe 2 parámetros que van a definir el tamaño del array que nos devuelve. Es decir si ponemos random.rand(2,2) nos va a devolver un array de 2 filas y 2 columnas (2x2). Los números aleatorios guardados en los arrays se escogen en un rango de 0 a 1.Vamos a poner un ejemplo de ello:

```python
array_aleatorio = np.random.rand(2,2)
array_aleatorio2 = np.random.rand(3,2)
print('Array 2x2')
print(array_aleatorio)
print('Array 3x2')
print(array_aleatorio2)
```
