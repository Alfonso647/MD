# Generación de grafos

En esta práctica vamos a estudiar los principios de los grafos sintéticos. Repasaremos tanto su creación como las propiedades de los mismos. Para esta práctica vamos a usar librerías las cuales se comentan más adelante, pero vamos a importarlas primeramente:

```python
import networkx as nx
import matplotlib.pyplot as plt
```

## 1. Grafos sintéticos

Un grafo sintético es un grafo generado de manera artificial, de manera que no se crea basándose de datos del mundo real. La utilidad de estos grafos está en que son perfectos para probar algoritmos (de búsqueda, por ejemplo) puesto que podemos controlar sus propiedades y estructura. Para crear estos grafos, usamos la librería `networkx`. 

### 1.1 Grafos aleatorios

Un grafo aleatorio consiste en un grafo cuyas aristas se generan de forma aleatoria (con ciertas restricciones). Se trata de uno de los grafos más simples de la familia de los grafos sintéticos. Existen una serie de modelos para generar estos grafos. En esta sección vamos a estudiar algunos de esos modelos y como funcionan

#### 1.1.1 Modelo Erdos-Rényi

Funcionamiento del modelo:

- Partimos de un número fijo de nodos
- Añadimos aristas entre pares de nodos aplicando probabilidad

La librería `networkx` nos ofrece una función para poder crear grafos aleatorios basados en este modelo: `erdos_renyi_graph()`

Dicha función recibe como parámetros:
- Número de nodos
- Probabilidad de arista entre pares 

Vamos a poner un ejemplo de un grafo aleatorio crado con este modelo (de cara a visualizar el grafo vamos a usar la librería `matplotlib`)

```python
# Usamos la función "erdos_renyi_graph" para generar un grafo basado en el modelo Eros-Rényi
grafo = nx.erdos_renyi_graph(15, 0.25)   # 10 nodos | Probabilidad de arista del 50%

# Mostramos el grafo por pantalla
plt.figure(figsize=(16,12))
plt.title('Grafo aleatorio (Modelo Erdos-Rényi)')

# Vamos a calcular la posiciones (nodos)
posiciones = nx.spring_layout(grafo, seed = 8)

# Dibujamos los nodos y arsitas
nx.draw_networkx_nodes(grafo, posiciones, node_color = 'black', node_size = 325)
nx.draw_networkx_edges(grafo, posiciones, edge_color = 'red', style = '-.')

plt.axis('off')
plt.show()
```

#### 1.1.2 Modelo Barabási-ALbert

Vamos a estudiar ahora el modelo `Barabási-Albert` el cual nuevamente es un modelo usado para generar grafos aleatorios. \
Funcionamiento del modelo:
- Partimos de un número de nodos (generalmente pocos)
- Se van añadiendo nuevos nodos con un mecanismo de unión por preferencias (los nodos con grados superiores tienen más probabilidad de recibir nuevas aristas)

La librería `networkx` nos ofrece una función para poder crear grafos aleatorios basados en este modelo: `barabasi_albert_graph()`

Dicha función recibe como parámetros:
- Número de nodos al finalizar
- Número de enlaces que va a tener cada nodo nuevo

Vamos a poner un ejemplo de un grafo aleatorio crado con este modelo (de cara a visualizar el grafo vamos a usar la librería `matplotlib`)

```python
# Usamos la función "barabasi_albert_graph" para generar un grafo basado en el modelo Barabási-Albert
grafo = nx.barabasi_albert_graph(9, 4, seed=10)   # 10 nodos | Nº enlaces por nodo

# Mostramos el grafo por pantalla
plt.figure(figsize=(16,12))
plt.title('Grafo aleatorio (Modelo Barabási-Albert)')

# Vamos a calcular la posiciones (nodos)
posiciones = nx.spring_layout(grafo, seed = 8)

# Dibujamos los nodos y arsitas
nx.draw_networkx_nodes(grafo, posiciones, node_color = 'red', node_size = 325)
nx.draw_networkx_edges(grafo, posiciones, edge_color = 'black', style = '--')

plt.axis('off')
plt.show()
```

### 1.2 Grafos de mundos pequeños

En este sección vamos a estudiar otra especie de grafo sintético: **Grafos de mundos pequeños**. Este tipo de grafo es un tipo de red compleja donde la mayoría de nodos no son vecinos directos, pero se puede llegar de un nodo a otro en un número bajo de pasos. 

La librería `NetworkX` nos ofrece una función para generar este tipo de grafo: ´watts_strogatz_graph()´, que recibe como parámetros:
- Número total de nodos que va a tener el grafo
- Grado medio de los nodos
- Probabilidad de reconexión de cada arsita (entre 0 y 1)

_Nota_ : _Esta función está basada en el modelo de Watts-Strogatz, también conocido como el modelo de los "seis grados de separación"_

Vamos a poner un ejemplo de un grafo aleatorio crado con este modelo (de cara a visualizar el grafo vamos a usar la librería `matplotlib`)

```python
# Usamos la función "watts_strogatz_graph" para generar un grafo basado en el modelo Watts-Strogatz
grafo = nx.watts_strogatz_graph(9, 4, 0.5)   # 12 nodos | Grado medio | Probabilidad reconexión

# Mostramos el grafo por pantalla
plt.figure(figsize=(16,12))
plt.title('Grafo aleatorio (Modelo Watts-Strogatz)')

# Vamos a calcular la posiciones (nodos)
posiciones = nx.spring_layout(grafo, seed = 16)

# Dibujamos los nodos y arsitas
nx.draw_networkx_nodes(grafo, posiciones, node_color = 'yellow', node_size = 500)
nx.draw_networkx_edges(grafo, posiciones, edge_color = 'green', style = '--')

plt.axis('off')
plt.show()
```

### 1.3 Grafos en árbol

Vamos a estudiar en este sección los grafos en árbol. Este tipo de grafo contiene la estructura más organizada y jerárquica que existe en la teoría de grafos. Su principal característica es que es acíclico (no tiene ciclos) y conexo (completamente conectado)

Características:
- Conexión única (solo ún camino entre cualquier par de nodos)
- Si un árbol tiene $X$ nodos, tendrá **siempre** $X-1$ aristas
- Se suelen dibujar de arriba hacia abajo 

`NetworkX` nos ofrece una función para generar grafos aleatorios de árbol: `random_tree()`, la cual recibe como parámetros:
- Número de nodos que tendrá el árbol
- Generador número aleatorios (semilla, _seed_)

Vamos a poner un ejemplo de un grafo aleatorio crado con este modelo (de cara a visualizar el grafo vamos a usar la librería `matplotlib`)

```python
# Usamos la función "random_tree" para generar un grafo de árbol
grafo = nx.random_tree(15, seed=99)   # 10 nodos

# Mostramos el grafo por pantalla
plt.figure(figsize=(16,12))
plt.title('Grafo aleatorio de árbol')

# Vamos a calcular la posiciones (nodos)
posiciones = nx.spring_layout(grafo, seed = 16)

# Dibujamos los nodos y arsitas
nx.draw_networkx_nodes(grafo, posiciones, node_color = 'skyblue', node_size = 500)
nx.draw_networkx_edges(grafo, posiciones, edge_color = 'orange', style = ':')

plt.axis('off')
plt.show()
```

### 1.4. Modelo de bloques estocástico (_SBM_)

Vamos a pasar a estuidar el modelo de bloques estocástico. Este modelo es un modelo generativo de grafos que captura la estructura comunitaria de grafos. Este modelo es utilizado para crear/estudiar redes que tienen comunidades o grupos definidos. Este modelo (al contrario que los anteriores) se basa en la pertenencia a grupos. 

Funcionamiento del modelo:
- Nodos se dividen en grupos, comunidades...
- Arsitas se generan entre nodos dependiendo del nodo al que pertenecen

`NetworkX` nos ofrece la función `stchastic_block_model` para generar este tipo de grafos. Dicha función recibe como parámetros:
- Lista con el número de nodos por comunidad
- Matriz de probabilidades de conexión
- Semilla para que el grafo sea reproducible

Vamos a poner un ejemplo de un grafo aleatorio crado con este modelo (de cara a visualizar el grafo vamos a usar la librería `matplotlib`)

```python
# Usamos la función "stochastic_block_modele" para generar un grafo aleatorio basado en SBM
grafo = nx.stochastic_block_model([20, 10, 5], [[0.6, 0.05, 0.05],
                                                [0.05, 0.6, 0.05], 
                                                [0.05, 0.05, 0.6]], seed=99)   # 3 comunidades

# Mostramos el grafo por pantalla
plt.figure(figsize=(16,12))
plt.title('Grafo aleatorio (SBM)')

# Vamos a calcular la posiciones (nodos)
posiciones = nx.spring_layout(grafo, seed = 16)

# Dibujamos los nodos y arsitas
nx.draw_networkx_nodes(grafo, posiciones, node_color = 'purple', node_size = 600)
nx.draw_networkx_edges(grafo, posiciones, edge_color = 'black', style = '-.')

plt.axis('off')
plt.show()
```

## 2. Propiedades de un grafo

En esta sección vamos a estudiar las diferentes propiedades que tienen los grafos. Estas propiedades nos sirven para obtener información útil de los grafos. Para acceder a estas propiedades vamos a usar las distintias función que nos ofrece `NetworkX`. A lo largo de esta sección pondremos un ejemplo de cara propiedad de cara a un resumen más claro.

#### &rarr; **Número de nodos y aristas**

Podemos obtener el número de nodos y arsitas de un grafo usando las siguientes funciones:
- number_of_nodes()
- number_of_edges()

Ejemplo:

```python
# Creación grafo
grafo = nx.random_tree(7, seed=99)   # 10 nodos
plt.figure(figsize=(10,5))
plt.title('Grafo aleatorio de árbol')
posiciones = nx.spring_layout(grafo, seed = 16)
nx.draw_networkx_nodes(grafo, posiciones, node_color = 'green', node_size = 500)
nx.draw_networkx_edges(grafo, posiciones, edge_color = 'orange', style = ':')
plt.axis('off')
plt.show()

# Vamos a obtener el número de nodos y aristas
nodos = nx.number_of_nodes(grafo)
aristas = nx.number_of_edges(grafo)

print(f'Número de nodos: {nodos}')
print(f'Número de aristas: {aristas}')
```

#### &rarr; **Es conexo**

Un grafo es conexo si existe un camino para cada par de nodos del grafo. Para comprobar si un grafo cumple esta propiedad, usamos la función:
- is_connected()

Ejemplo:

```python
# Creación grafo
grafo = nx.watts_strogatz_graph(9, 4, 0.5)   # 12 nodos | Grado medio | Probabilidad reconexión
plt.figure(figsize=(10,5))
plt.title('Grafo aleatorio (Modelo Watts-Strogatz)')
posiciones = nx.spring_layout(grafo, seed = 16)
nx.draw_networkx_nodes(grafo, posiciones, node_color = 'yellow', node_size = 500)
nx.draw_networkx_edges(grafo, posiciones, edge_color = 'green', style = '--')
plt.axis('off')
plt.show()

# Comprobamos si el grafo es conexo
conexo: bool = nx.is_connected(grafo)

print(f'¿És conexo el grafo?: {'Sí' if conexo else }')
```

#### &rarr; **No tiene ciclos**

Un grafo es acíclico si no contiene ningún ciclo. Podemos comprobar si un grafo cumple esa propiedad con la función:
- is_directed_acyclic_grpah()

Ejemplo:

```python
# Creación grafo
grafo = nx.random_tree(15, seed=99)   # 10 nodos
plt.figure(figsize=(10,5))
plt.title('Grafo aleatorio de árbol')
posiciones = nx.spring_layout(grafo, seed = 16)
nx.draw_networkx_nodes(grafo, posiciones, node_color = 'yellow', node_size = 500)
nx.draw_networkx_edges(grafo, posiciones, edge_color = 'orange', style = ':')
plt.axis('off')
plt.show()

# Comprobamos si el grafo es acíclico 

ciclico: bool = nx.is_directed_acyclic_graph(grafo)

print(f'¿És cíclico el grafo?: {'Sí' if ciclico else 'No'}')
```

#### &rarr; **Radio y diámetro**

- El diámetro es la distancia más grande que existe entre dos puntos de la red (máximo trayecto necesario para cruzar el grafo de un extremo a otro por el camino más corto): \
`radius()`

- El radio mide la distancia máxima desde el nodo que está mejor ubicado hasta el punto más alejado él: \
`diameter()`

Ejemplo:

```python
# Creación grafo
plt.figure(figsize=(10,5))
plt.title('Grafo aleatorio (Modelo Barabási-Albert)')
posiciones = nx.spring_layout(grafo, seed = 8)
nx.draw_networkx_nodes(grafo, posiciones, node_color = 'blue', node_size = 325)
nx.draw_networkx_edges(grafo, posiciones, edge_color = 'black', style = '--')
plt.axis('off')
plt.show()

# Vamos a calcular el diámetro y radio del grafo
diametro = nx.diameter(grafo)
radio = nx.radius(grafo)

print(f'Diámetro del grafo: {diametro}')
print(f'Radio del grafo: {radio}')
```

#### &rarr; **Grado medio**

El grado medio de un grafo es el promedio de conexión que tiene cada nodo. Podemos calcular este dato usando la siguiente función:
- `average_degree_connectivity()`

Ejemplo:

```python
# Creación grafo
grafo = nx.watts_strogatz_graph(12, 4, 0.5)
plt.figure(figsize=(10,5))
plt.title('Grafo aleatorio (Modelo Watts-Strogatz)')
posiciones = nx.spring_layout(grafo, seed = 16)
nx.draw_networkx_nodes(grafo, posiciones, node_color = 'green', node_size = 500)
nx.draw_networkx_edges(grafo, posiciones, edge_color = 'green', style = ':')
plt.axis('off')
plt.show()

# Calculamos el grado medio del grafo
g_medio = nx.average_degree_connectivity(grafo)

print(f'Grado medio del grafo: {g_medio}')
```

#### &rarr; **Distribución de grados**

La distribución de grados se refiere a la distribución de probabilidad de los grados de los nodos en el grafo. Para calcular este dato usamos la siguente función:
- degree_histogram()

Ejemplo:

```python
# Creación grafo
grafo = nx.stochastic_block_model([20, 10, 5], [[0.6, 0.05, 0.05],
                                                [0.05, 0.6, 0.05], 
                                                [0.05, 0.05, 0.6]], seed=99)   # 3 comunidades
plt.figure(figsize=(10,5))
plt.title('Grafo aleatorio (SBM)')
posiciones = nx.spring_layout(grafo, seed = 16)
nx.draw_networkx_nodes(grafo, posiciones, node_color = 'skyblue', node_size = 600)
nx.draw_networkx_edges(grafo, posiciones, edge_color = 'green', style = '-.')

plt.axis('off')
plt.show()

# Obtenemos la distribución de grados
dist = nx.degree_histogram(grafo)

print(f'Distribución probabilidad: {dist}')
```

#### &rarr; **Longitud media del camino más corto**

La longitud media del camino más corto en un grafo se refiere a la distancia media que tiene los partes de nodos. Para calcular este dato, tenemos la función:
- `average_shortest_path_length()`

Ejemplo:

```python
# Creación de grafo
grafo = nx.watts_strogatz_graph(11, 5, 0.5)  
# Mostramos el grafo por pantalla
plt.figure(figsize=(16,12))
plt.title('Grafo aleatorio (Modelo Watts-Strogatz)')
posiciones = nx.spring_layout(grafo, seed = 16)
nx.draw_networkx_nodes(grafo, posiciones, node_color = 'blue', node_size = 500)
nx.draw_networkx_edges(grafo, posiciones, edge_color = 'purple', style = '--')

plt.axis('off')
plt.show()

# Obtenemos la longitud media del camino más corto
l_media = nx.average_shortest_path_length(grafo)

print(f'Longitud media: {l_media}')
```

### Ejercicio 2

```python
# Creamos 4 bloques de 10 nodos cada uno
sizes = [10, 10, 10, 10]

# Definimos las matrices de probabilidad:
    # Baja: Probabilidad baja conexión entre bloques distintos
    # Alta: Probabilidad alta conexión entre bloques distintos
baja = [
    [0.8, 0.02, 0.02, 0.02],
    [0.02, 0.8, 0.02, 0.02],
    [0.02, 0.02, 0.8, 0.02],
    [0.02, 0.02, 0.02, 0.8]
]

alta = [
    [0.8, 0.8, 0.8, 0.8],
    [0.8, 0.8, 0.8, 0.8],
    [0.8, 0.8, 0.8, 0.8],
    [0.8, 0.8, 0.8, 0.8]
]

# Definimos la semilla para que los resultados sean reproducibles
seed = 96

# Creamos los 2 grafos (SBM)
G1 = nx.stochastic_block_model(sizes, baja, seed=seed)  # El primero -> Probabilidad alta
G2 = nx.stochastic_block_model(sizes, alta, seed=seed)  # El segundo -> Probabilidad baja

# Calculamos las posiciones (las mismas para los dos grafos)
pos1 = nx.spring_layout(G1, seed=seed)
pos2 = nx.spring_layout(G2, seed=seed)

# Vamos a dividir los dos grafos en una misma figura
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Primer grafo
#--------------#
# Definimos título
axes[0].set_title('SBM - Baja conexion entre bloques')  
nx.draw_networkx_nodes(G1, pos1, node_size=300, node_color='green', ax=axes[0])  # Dibujamos los nodos
nx.draw_networkx_edges(G1, pos1, width=1.0, alpha=0.5, edge_color='black', style='-.', ax=axes[0]) # Dibujamos las aristas
axes[0].axis('off')

# Segundo grafo
#---------------#
# Definimos título
axes[1].set_title('SBM - Alta conexion entre bloques')
nx.draw_networkx_nodes(G2, pos2, node_size=300, node_color= 'yellow', ax=axes[1])         # Dibujamos los nodos
nx.draw_networkx_edges(G2, pos2, width=1.0, alpha=0.5, style=':', ax=axes[1])  # Dibujamos las arsitas
axes[1].axis('off')

# Mostramos el resultado por pantalla
plt.show()
```

En el primer grafo (izquierda), cada comunidad está muy bien cohesionada internamente pero casi aislada del resto. Si imaginamos que cada bloque es un grupo de personas (por ejemplo, departamentos de una empresa), comunicarse con alguien de otro departamento sería muy difícil: habría pocos puentes entre grupos, los caminos serían largos y pasarían por muy pocos nodos intermedios, creando cuellos de botella. Una información que salga de un bloque tardará mucho en llegar a otro.
En el segundo grafo, todos los nodos se conectan con alta probabilidad independientemente de su bloque. La información fluye rápidamente entre cualquier par de nodos, el diámetro del grafo es mucho menor y no existen grupos aislados. Sería mucho más fácil comunicarse: cualquier persona puede llegar a otra en pocos pasos. Sin embargo, se pierde la estructura de comunidad, ya que los bloques dejan de ser distinguibles visualmente.
