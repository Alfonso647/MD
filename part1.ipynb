## 1. Grafos sintéticos

Un grafo sintético es un grafo generado de manera artificial, de manera que no se crea basándose de datos del mundo real. La utilidad de estos grafos está en que son perfectos para probar algoritmos (de búsqueda, por ejemplo) puesto que podemos controlar sus propiedades y estructura. Para crear estos grafos, usamos la librería `networkx`. 

### 1.1 Grafos aleatorios

Un grafo aleatorio consiste en un grafo cuyas aristas se generan de forma aleatoria (con ciertas restricciones). Se trata de uno de los grafos más simples de la familia de los grafos sintéticos. Existen una serie de modelos para generar estos grafos. En esta sección vamos a estudiar algunos de esos modelos y como funcionan

#### 1.1.1 Modelo Erdos-Rényi

Funcionamiento del modelo:

- Partimos de un número fijo de nodos
- Añadimos aristas entre pares de nodos aplicando probabilidad

La librería `networkx` nos ofrece una función para poder crear grafos aleatorios basados en este modelo: `erdos_renyi_graph()`

Dicha función recibe como parámetros:
- Número de nodos
- Probabilidad de arista entre pares 

Vamos a poner un ejemplo de un grafo aleatorio crado con este modelo (de cara a visualizar el grafo vamos a usar la librería `matplotlib`)

```python
# Usamos la función "erdos_renyi_graph" para generar un grafo basado en el modelo Eros-Rényi
grafo = nx.erdos_renyi_graph(15, 0.25)   # 10 nodos | Probabilidad de arista del 50%

# Mostramos el grafo por pantalla
plt.figure(figsize=(16,12))
plt.title('Grafo aleatorio (Modelo Erdos-Rényi)')

# Vamos a calcular la posiciones (nodos)
posiciones = nx.spring_layout(grafo, seed = 8)

# Dibujamos los nodos y arsitas
nx.draw_networkx_nodes(grafo, posiciones, node_color = 'black', node_size = 325)
nx.draw_networkx_edges(grafo, posiciones, edge_color = 'red', style = '-.')

plt.axis('off')
plt.show()
```
