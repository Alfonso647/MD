## 1. Creación de DataFrames

Un DataFrame es ... Vamos a estudiar diferentes formas de crear DataFrames.

### 1.1 Creación de DataFrame a partir de un diccionario

Pandas nos ofrece la opción de crear un DataFrame con un diccionario usando la función `pd.DataFrame()`. Funciona de la siguiente forma:
- Las keys del diccionario se refieren a los nombres de las columnas
- Los values de dichas keys son los valores de dichas columnas

Vamos a crear un DataFrame de ejemplo usando este método:

````python
# Creamos el diccionario
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Emily'],
    'Age': [25, 30, 35, 40, 45],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix']
}

# Creamos el DataFrame (usando el diccionario)
dataframe = pd.DataFrame(data)

print(dataframe)
````

### 1.2 Creación de un DataFrame a partir de una lista de diccionarios

Otra forma de crear DataFrame es empleando una lista que contiene diccionarios. Para crearlo, primero definimos dicha lista y volvemos a usar la función `pd.DataFrame()`. En este caso, la distribución es la siguiente:
- Cada diccionario de la lista es una fila del DataFrame

Vamos a crear un DataFrame de ejemplo usando este método:

````python
# Creamos la lista de diccionarios
data = [
    {'Name': 'Alice', 'Age': 25, 'City': 'New York'},
    {'Name': 'Bob', 'Age': 30, 'City': 'Los Angeles'},
    {'Name': 'Charlie', 'Age': 35, 'City': 'Chicago'},
    {'Name': 'David', 'Age': 40, 'City': 'Houston'},
    {'Name': 'Emily', 'Age': 45, 'City': 'Phoenix'}
]

# Creamos el DataFrame (usando la lista de diccionarios)
dataframe = pd.DataFrame(data)

print(dataframe)
````

### 1.3 Creación de un DataFrame a partir de una lista de listas

Pandas también permite crear DataFrames usando una lista de listas. Nuevamente, tenemos que definir dicha lista y usar la función `pd.DataFrame()`. En este caso, el DataFrame se ordena de la siguiente forma:
- Cada lista se refiere a una fila del DataFrame

Vamos a crear un DataFrame de ejemplo usando este método:

````python
# Creamos la lista de listas

data = [
    ['Alice', 25, 'New York'],
    ['Bob', 30, 'Los Angeles'],
    ['Charlie', 35, 'Chicago'],
    ['David', 40, 'Houston'],
    ['Emily', 45, 'Phoenix']
]

# Creamos el DataFrame (usando la lista de listas)
dataframe = pd.DataFrame(data, columns=['Name', 'Age', 'City'])

print(dataframe)
````

En nuesto caso hemos usado crear al DataFrame con una matriz Numpy

## 2. Acceso a DataFrames

Tenemos varias formas acceder a los DataFrames. Estas nos permiten acceder de manera individual y conjunta a filas, columnas o celdas. En este apartado vamos a estudiarlas en profundidad

### 2.1 Acceso a columnas individuales

Para acceder a columnas individuales de un DataFrame, usamos el nombre de dicha columna. Vamos a ver un ejemplo

````python
print(df)
print('\nColumna que contiene los nombres de los planetas y satélites')
entes = df['Nombre']
print(entes)
````

### 2.2 Acceso a filas individuales

La forma de acceso a filas de un DataFrame es usando el método `iloc[]`. Este método recibe como parámetro el índice de la fila a la que queramos acceder. Vamos a ver un ejemplo:

````python
print(df)
print('\nFila con la información del planeta Tierra')
Tierra = df.iloc[3]
print(Tierra)
````

### 2.3 Acceso a celdas individuales

Para el acceso a una celda, necesitamos especificar la fila y columna donde esta se encuentra. Luego para realizar este acceso necesitamos lo siguiente:
- El método `iloc[]` -> Especificamos la fila
- El nombre de la columna -> Especificamos la columna

````python
print(df)
print('\nRadio de Saturno')
radio_saturno = df.iloc[6]['Radio']
print(f'{radio_saturno} km')
````

## 3. Operaciones en DataFrames

Una vez definido un DataFrame, podemos realizar un conjunto de operaciones sobre este con diversos objetivos como filtrar datos, agruparlos, agregarlos... En este apartado vamos a estudiar dichas operaciones en profundidad

### 3.1 Filtrado de datos

Podemos filtar filas que cumplan una condición definida para realizar un filtrado de datos sobre nuestro DataFrame. Vamos a ver un ejemplo:

````python
print(df)
print('\nPlanetas con radio > 20000 km')
planetas = df[(df['Tipo'] == 'Planeta') & (df['Radio'] > 20000)]
print(planetas)
````

### 3.2 Ordenar datos DataFrame

Otra funcioliadad disponible es ordenar nuestro DataFrame. Podemos hacerlo por una o varias columnas. Para ello, usamos la función `sort_values()`. Esta función recibe los parámetros:
- by='' -> Especificamos por cual columna/fila queremos que se ordene con su nombre
- ascending: bool - Especificamos si queremos que se ordene de forma ascendente o descendente.

Vamos a ver varios ejemplos con nuestro df:

````python
print(f'DataFrame original: \n{df}\n-------------------------------------------')
ord_radio = df.sort_values(by='Radio', ascending=True)
print(f'DataFrame ordenado por "Radio" (ascendente):\n{ord_radio}')
````

Vamos a poner otro ejemplo ordenando ahora de forma descendente:

````python
print(f'DataFrame original: \n{df}\n-------------------------------------------')
ord_radio_dist = df.sort_values(by=['ID', 'Radio'], ascending=False)
print(f'DataFrame ordenado por "ID" (descendente):\n{ord_radio_dist}')
````

### 3.3 Estadísitcas sobre DataFrames

Podemos obtener información acerca de nuestros DataFrames empleando sobre ellos operaciones estadísticas, como la media, mínima, máxima... En este apartado vamos a esutidar como realizarlo. V

Funciones:
1. mean() -> media
2. min() -> valor mínimo
3. max() -> valor máximo
4. medain() -> mediana
5. sum() -> suma total
etc 

Vamos a reliazar estas operaciones sobre nuestro DataFrame a modo de ejemplo:


````python
print('INFO DE NUESTRO DATA FRAME\n----------------------------')

media = df['Radio'].mean()
print(f'Media de los radios: {media}')
min_radio = df['Radio'].min()
print(f'Menor radio del sistema solar: {min_radio}')
max_dist = df['Distancia_UA'].max()
print(f'Mayor distancia de un planeta con respecto a la UA: {max_dist}')
mediana = df['Radio'].median()
print(f'Mediana de los radios:{mediana}')
suma = df['Radio'].sum()
print(f'Suma total de los radios: {suma}')
````
