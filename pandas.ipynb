```python
import pandas as pd
import numpy as np

# Create a NumPy array
datos = np.array([
    ['Alice', 25, 'New York'],
    ['Bob', 30, 'Los Angeles'],
    ['Charlie', 35, 'Chicago'],
    ['David', 40, 'Houston'],
    ['Emily', 45, 'Phoenix']
])

# Create a DataFrame from the NumPy array
df = pd.DataFrame(datos, columns=['Name', 'Age', 'City'])

# Display the DataFrame
print(df)
# Guardamos el dataframe
df.to_csv('datos.csv', index=False)
```

## 1. Creación de DataFrames

Un DataFrame es ... Vamos a estudiar diferentes formas de crear DataFrames.

### 1.1 Creación de DataFrame a partir de un diccionario

Pandas nos ofrece la opción de crear un DataFrame con un diccionario usando la función `pd.DataFrame()`. Funciona de la siguiente forma:
- Las keys del diccionario se refieren a los nombres de las columnas
- Los values de dichas keys son los valores de dichas columnas

Vamos a crear un DataFrame de ejemplo usando este método:

```python
# Creamos el diccionario
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Emily'],
    'Age': [25, 30, 35, 40, 45],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix']
}

# Creamos el DataFrame (usando el diccionario)
df = pd.DataFrame(data)

print(df)
```

### 1.2 Creación de un DataFrame a partir de una lista de diccionarios

Otra forma de crear DataFrame es empleando una lista que contiene diccionarios. Para crearlo, primero definimos dicha lista y volvemos a usar la función `pd.DataFrame()`. En este caso, la distribución es la siguiente:
- Cada diccionario de la lista es una fila del DataFrame

Vamos a crear un DataFrame de ejemplo usando este método:

```python
# Creamos la lista de diccionarios
data = [
    {'Name': 'Alice', 'Age': 25, 'City': 'New York'},
    {'Name': 'Bob', 'Age': 30, 'City': 'Los Angeles'},
    {'Name': 'Charlie', 'Age': 35, 'City': 'Chicago'},
    {'Name': 'David', 'Age': 40, 'City': 'Houston'},
    {'Name': 'Emily', 'Age': 45, 'City': 'Phoenix'}
]

# Creamos el DataFrame (usando la lista de diccionarios)
df = pd.DataFrame(data)

print(df)
```

### 1.3 Creación de un DataFrame a partir de una lista de listas

Pandas también permite crear DataFrames usando una lista de listas. Nuevamente, tenemos que definir dicha lista y usar la función `pd.DataFrame()`. En este caso, el DataFrame se ordena de la siguiente forma:
- Cada lista se refiere a una fila del DataFrame

Vamos a crear un DataFrame de ejemplo usando este método:

```python
# Creamos la lista de listas

data = [
    ['Alice', 25, 'New York'],
    ['Bob', 30, 'Los Angeles'],
    ['Charlie', 35, 'Chicago'],
    ['David', 40, 'Houston'],
    ['Emily', 45, 'Phoenix']
]

# Creamos el DataFrame (usando la lista de listas)
df = pd.DataFrame(data, columns=['Name', 'Age', 'City'])

print(df)
```

En nuesto caso hemos usado crear al DataFrame con una matriz Numpy
